```python
import mysql.connector

mydb = mysql.connector.connect(user='root', password='root', host='localhost', database = 'db')

c = mydb.cursor(buffered = True)

c.execute("CREATE DATABASE db")

c.execute("USE db")
```

```python
c.execute("SHOW DATABASES")

for x in c:
    print(x)
```

```python
c.execute("CREATE TABLE customers (name VARCHAR(255), address VARCHAR(255))")
```

```python
try:
    c.fetchall()
except:
    pass

c.execute("SHOW TABLES")

for x in c:
    print(x)
```

```python
c.execute("ALTER TABLE customers ADD COLUMN id INT AUTO_INCREMENT PRIMARY KEY")
```

```python
sql = "INSERT INTO customers (name, address) VALUES (%s, %s)"
val = ("John", "Highway 21")
c.execute(sql, val)
```

```python
sql = "DELETE FROM customers WHERE address LIKE %s"
addr = ('Ocean blvd 2', )
c.execute(sql, addr)
mydb.commit()
print('record delete')
```

```python
sql = "UPDATE customers SET address = 'Canyon 123' WHERE address = 'Valley 345'"
c.execute(sql)
mydb.commit()
print(c.rowcount,'record updated')
```

```python
sql = "UPDATE customers SET address = %s WHERE address = %s"
addr = ("Valley 345","Canyon 123")
c.execute(sql, addr)
mydb.commit()
print(c.rowcount,'record updated')
```
